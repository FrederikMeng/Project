```python
def day_of_week(day, month, year):
    month_codes = [0, 3, 3, 6, 1, 4, 6, 2, 5, 0, 3, 5]
    newmonth = month_codes[month - 1]

    century = year // 100      # 1999 -> 19
    short_year = year % 100    # 1999 -> 99

    century_code = (2 * (3 - century % 4)) % 7

    calc = short_year + short_year // 4 + day + newmonth + century_code

    is_leap = (year % 4 == 0 and year % 100 != 0) or year % 400 == 0
    if is_leap and month <= 2:
        calc = calc - 1

    days = ["sunday", "monday", "tuesday", "wednesday", "thursday", "friday", "saturday"]
    return days[calc % 7]


def birthday():
    birth_date = input("Write your birthday as day month year, separated by spaces (e.g. 24 12 1999): ")
    day, month, year = birth_date.split()   # only works with or without leading zeros, "05" or "5"
    day = int(day)
    month = int(month)
    year = int(year)
    return "you were born on a " + day_of_week(day, month, year)


print(birthday())
```
